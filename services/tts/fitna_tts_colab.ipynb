### خطوات التشغيل:
1. تأكد من تفعيل كارت الشاشة: من القائمة العلوية **Runtime** -> **Change runtime type** -> اختر **T4 GPU** -> اضغط **Save**.
2. اضغط على **Runtime** -> **Run all** (أو اضغط `Ctrl + F9`).
3. في نهاية الخلية الأخيرة، ستجد رابطاً مثل: `https://xxxx.trycloudflare.com`.
4. انسخ هذا الرابط وضعه في ملف `.env.local` في مشروعك بهذا الشكل:
```env
LOCAL_TTS_URL=https://xxxx.trycloudflare.com/tts
```
5. مبروك! طلاب الفصل سيتحدثون فوراً بأصواتهم الحقيقية وبسرعة فائقة!

````python
# 3. استخراج الأصوات المرجعية للطلاب الأربعة تلقائياً
import base64, os
os.makedirs("voices", exist_ok=True)

voices = {
  "omar.wav": "",
  "sara.wav": "",
  "yassin.wav": "",
  "nour.wav": ""
}

for filename, b64 in voices.items():
    with open(f"voices/{filename}", "wb") as f:
        f.write(base64.b64decode(b64))
    print(f"[✓] تم تجهيز صوت: {filename}")
````

````python
# 4. تشغيل سيرفر XTTS v2 التوليدي
import io, torch, numpy as np, threading, time, subprocess, re
from fastapi import FastAPI, HTTPException, Response
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
import scipy.io.wavfile as wavfile
import uvicorn

# إصلاح توافق مكتبة transformers
try:
    import transformers.pytorch_utils
    if not hasattr(transformers.pytorch_utils, "isin_mps_friendly"):
        transformers.pytorch_utils.isin_mps_friendly = torch.isin
except Exception:
    pass

from TTS.api import TTS

app = FastAPI(title="Fitna AI TTS")
app.add_middleware(CORSMiddleware, allow_origins=["*"], allow_credentials=True, allow_methods=["*"], allow_headers=["*"])

print("[*] جاري تحميل موديل XTTS v2 على كارت الشاشة...")
tts_model = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to("cuda")
print("[+] الموديل جاهز تماماً للاستنساخ!")

class TTSReq(BaseModel):
    text: str
    persona: str = "omar"
    language: str = "ar"

@app.post("/tts")
def synthesize(req: TTSReq):
    persona = req.persona.lower().strip()
    name_map = {"عمر": "omar", "سارة": "sara", "ياسين": "yassin", "نور": "nour"}
    key = name_map.get(persona, persona)
    ref = f"voices/{key}.wav"
    if not os.path.exists(ref):
        ref = "voices/omar.wav"
    
    wav = tts_model.tts(text=req.text.strip(), speaker_wav=ref, language="ar")
    arr = (np.array(wav, dtype=np.float32) * 32767).clip(-32768, 32767).astype(np.int16)
    buf = io.BytesIO()
    wavfile.write(buf, 24000, arr)
    buf.seek(0)
    return Response(content=buf.read(), media_type="audio/wav")

def start_uvicorn():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

threading.Thread(target=start_uvicorn, daemon=True).start()
time.sleep(3)

print("[*] جاري إنشاء الرابط السحابي المباشر عبر Cloudflare...")
tunnel_proc = subprocess.Popen(["cloudflared", "tunnel", "--url", "http://127.0.0.1:8000"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
for line in tunnel_proc.stdout:
    match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
    if match:
        public_url = match.group(0)
        print("\n=======================================================")
        print("🎉 تم إنشاء الرابط بنجاح!")
        print(f"انسخ الرابط التالي وضعه في ملف .env.local بمشروعك:")
        print(f"LOCAL_TTS_URL={public_url}/tts")
        print("=======================================================\n")
        break

while True:
    time.sleep(3600)
````
